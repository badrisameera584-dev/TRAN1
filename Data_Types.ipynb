```python
class Employee:
    def __init__(self, employee_id, employee_name, age, salary):
        self.employee_id = employee_id
        self.employee_name = employee_name
        self.age = age
        self.salary = salary

    def display(self):
        print(f"Employee ID: {self.employee_id}")
        print(f"Employee Name: {self.employee_name}")
        print(f"Age: {self.age}")
        print(f"Salary: {self.salary}")


# Create employee 1 using user input
emp1_id = int(input("Enter Employee ID: "))
emp1_name = input("Enter Employee Name: ")
emp1_age = int(input("Enter Employee Age: "))
emp1_salary = float(input("Enter Employee Salary: "))
emp1 = Employee(emp1_id, emp1_name, emp1_age, emp1_salary)

# Create employee 2 using user input
emp2_id = int(input("Enter Employee ID: "))
emp2_name = input("Enter Employee Name: ")
emp2_age = int(input("Enter Employee Age: "))
emp2_salary = float(input("Enter Employee Salary: "))
emp2 = Employee(emp2_id, emp2_name, emp2_age, emp2_salary)

# Display both employees' details
emp1.display()
emp2.display()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Create a small DataFrame with numeric features for correlation analysis
data = {
    'study_hours': [2, 4, 6, 8, 10, 12, 14, 16],
    'marks': [40, 55, 60, 70, 80, 85, 90, 95],
    'attendance': [60, 70, 75, 80, 85, 90, 92, 95],
}
df = pd.DataFrame(data)

corr = df.corr()

plt.figure(figsize=(8, 6))
plt.imshow(corr, cmap='coolwarm', vmin=-1, vmax=1)
plt.xticks(range(len(corr.columns)), corr.columns, rotation=45)
plt.yticks(range(len(corr.columns)), corr.columns)
for i in range(corr.shape[0]):
    for j in range(corr.shape[1]):
        plt.text(j, i, f"{corr.iloc[i, j]:.2f}", ha='center', va='center')
plt.title('Correlation Heatmap of Numeric Features')
plt.colorbar()
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import pandas as pd
from scipy import stats

# Probability
prob_marks_above_80 = (df["Marks"] > 80).mean()
prob_sleep_7_plus = (df["Sleep_Hours"] >= 7).mean()
print("Probability Marks > 80:", prob_marks_above_80)
print("Probability Sleep >= 7 hours:", prob_sleep_7_plus)

# Vectors
marks_vector = df["Marks"].to_numpy()
study_vector = df["Study_Hours"].to_numpy()
print("Marks vector head:", marks_vector[:5])
print("Study vector head:", study_vector[:5])
print("Vector addition head:", marks_vector[:5] + study_vector[:5])
print("Dot product:", np.dot(marks_vector[:5], study_vector[:5]))

# Class and Objects
class Student:
    def __init__(self, name, marks, sleep_hours):
        self.name = name
        self.marks = marks
        self.sleep_hours = sleep_hours

    def performance(self):
        if self.marks >= 80:
            return "Excellent"
        elif self.marks >= 60:
            return "Good"
        return "Needs Improvement"

    def summary(self):
        return f"{self.name}: marks={self.marks}, sleep={self.sleep_hours}, status={self.performance()}"

student1 = Student("Asha", 88, 7)
print(student1.summary())

# Correlation
correlation_matrix = df[["Study_Hours", "Sleep_Hours", "Marks"]].corr()
print(correlation_matrix)
print("Correlation Study Hours vs Marks:", df["Study_Hours"].corr(df["Marks"]))
print("Correlation Sleep Hours vs Marks:", df["Sleep_Hours"].corr(df["Marks"]))

# Feature Intuition
avg_marks_by_sleep = df.groupby("Sleep_Hours")["Marks"].mean()
print(avg_marks_by_sleep)

# Loops
for col in df.columns:
    print(col, "->", df[col].dtype)

for i in range(3):
    print("Row", i, df.iloc[i].to_dict())

# Matrix
matrix = np.array([[2, 1, 0], [1, 3, 1], [0, 1, 2]])
print("Matrix:\n", matrix)
print("Transpose:\n", matrix.T)
print("Determinant:", np.linalg.det(matrix))
print("Inverse:\n", np.linalg.inv(matrix))
```
